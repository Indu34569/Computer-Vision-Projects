```python
img = cv2.imread("images/Flower.jpeg")

horizontal=cv2.flip(img,1)
vertical=cv2.flip(img,0)
both=cv2.flip(img,-1)

plt.figure(figsize=(10,5))
plt.subplot(3,4,1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title("original")

plt.subplot(3,4,2)
plt.imshow(cv2.cvtColor(horizontal,cv2.COLOR_BGR2RGB))
plt.title("horizontal")

plt.subplot(3,4,3)
plt.imshow(cv2.cvtColor(vertical,cv2.COLOR_BGR2RGB))
plt.title("vertical")

plt.subplot(3,4,4)
plt.imshow(cv2.cvtColor(both,cv2.COLOR_BGR2RGB))
plt.title("both")

plt.show()
```

```python
import matplotlib.pyplot as plt
import cv2
img = cv2.imread("images/Flower.jpeg")
print("imageshape",img.shape)
h,w=img.shape[:2]
row_start=int(h*0.2)
row_end=int(h*0.6)
col_start=int(h*0.2)
col_end=int(h*0.6)
cropped=img[row_start:row_end,col_start:col_end]

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title("original")

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(cropped,cv2.COLOR_BGR2RGB))
plt.title("cropped")
```

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
img = cv2.imread("images/Flower.jpeg")
print("imageshape",img.shape)
rows,cols=img.shape[:2]
M=np.float32([[1,0,60],[0,1,20]])
translated=cv2.warpAffine(img,M,(cols,rows))
plt.imshow(cv2.cvtColor(translated,cv2.COLOR_BGR2RGB))
plt.title("translated")
plt.show()
```

```python
import matplotlib.pyplot as plt
import cv2
import numpy as np
img = cv2.imread("images/Flower.jpeg")
center=(cols//2,rows//2)
M=cv2.getRotationMatrix2D(center,45,1)
rotated=cv2.warpAffine(img,M,(cols,rows))
plt.imshow(cv2.cvtColor(rotated,cv2.COLOR_BGR2RGB))
plt.title("rotated")
plt.show()
```

```python

import matplotlib.pyplot as plt
import cv2
img = cv2.imread('images/Flower.jpeg')
cv2.rectangle(    img,    (50, 50),      # top-left    
                  (300, 200),    # bottom-right    
                  (0, 255, 0),   # Green (BGR)    
                  3              # Thickness 
             )
cv2.circle(    img,    (500, 150),    # Center    
               80,            # Radius    
               (255, 0, 0),   # Blue (BGR)    
               3              # Thickness 
          )
# Add Text 
cv2.putText(    img,    "Flower",    (150, 150),                    # Position   
                cv2.FONT_HERSHEY_SIMPLEX,      # Font    
                1,                             # Font Scale    
                (0, 0, 255),                   # Red (BGR)    
                2,                             # Thickness    
                cv2.LINE_AA                    # Smooth Text 
           ) 
# Display Image 
plt.figure(figsize=(10,6)) 
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)) 
plt.axis('off') 
plt.title("Rectangle + Circle + Text") 
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Create a blank white image (500x500)
img = np.ones((500, 500, 3), dtype=np.uint8) * 255

# Draw a rectangle
cv2.rectangle(img, (50, 50), (200, 200), (255, 0, 0), 3)

# Draw a circle
cv2.circle(img, (350, 150), 70, (0, 255, 0), 3)

# Add your name
cv2.putText(img,
            "INDU R",          # Replace with your name
            (120, 400),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 0, 255),
            2,
            cv2.LINE_AA)

# Display the image
plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("My Image Design")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Create a blank white image
img = np.ones((500, 300, 3), dtype=np.uint8) * 255

# Draw the traffic signal box
cv2.rectangle(img, (90, 40), (210, 320), (0, 0, 0), 3)

# Draw Red light
cv2.circle(img, (150, 90), 30, (0, 0, 255), -1)

# Draw Yellow light
cv2.circle(img, (150, 180), 30, (0, 255, 255), -1)

# Draw Green light
cv2.circle(img, (150, 270), 30, (0, 255, 0), -1)

# Add title
cv2.putText(img,
            "Traffic Signal",
            (35, 380),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (255, 0, 0),
            2,
            cv2.LINE_AA)

# Display the image
plt.figure(figsize=(4, 7))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.title("Traffic Signal")
plt.show()
```
