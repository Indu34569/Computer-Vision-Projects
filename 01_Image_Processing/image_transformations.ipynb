```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
img = cv2.imread("images/forest.jpeg")
```

```python
cv2.imshow("Image",img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
plt.imshow(img)
plt.title("BGR")
plt.axis('off')
plt.show()
```

```python
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis('off')
plt.show()
```

```python
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2GRAY))
plt.title("Gray")
plt.axis('off')
plt.show()
```

```python
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
plt.imshow(gray,cmap='gray')
plt.title("RGB Image")
plt.axis('off')
plt.show
```

```python
cv2.imshow("Image",gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
resized_img=cv2.resize(img,(300,300))
cv2.imshow("Image",resized_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
resized_img=cv2.resize(img,(300,300))
plt.figure(figsize=(20,5))
plt.subplot(1,3,1)
plt.imshow(cv2.cvtColor(resized_img,cv2.COLOR_BGR2RGB))
plt.title("original")
plt.show()
```

```python
scaled_img=cv2.resize(img,None,fx=0.2,fy=0.2)
scaled_rgb=cv2.cvtColor(scaled_img,cv2.COLOR_BGR2RGB)
plt.subplot(1,3,1)
plt.imshow(scaled_rgb)
plt.title("scaled_rgb")
```

```python
resized_img=cv2.resize(img,(300,300))

img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
resized_rgb=cv2.cvtColor(resized_img,cv2.COLOR_BGR2RGB)
scaled_rgb=cv2.cvtColor(scaled_img,cv2.COLOR_BGR2RGB)


plt.figure(figsize=(10,5))
plt.subplot(1,3,1)
plt.imshow(img_rgb)
plt.title("original")
plt.subplot(1,3,2)
plt.imshow(resized_rgb)
plt.title("resized_rgb")
plt.subplot(1,3,3)
plt.imshow(scaled_rgb)
plt.title("scaled_rgb")
plt.show()
```
