```python
import cv2
from ultralytics import YOLO

# Load YOLO model
model = YOLO("yolov8n.pt")

# Open webcam
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if not ret:
        break

    resized_frame = cv2.resize(frame, (1080, 600))

    # Perform detection
    results = model(resized_frame)

    # Draw bounding boxes
    annotated_frame = results[0].plot()

    cv2.imshow("YOLO Live Detection", annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
import os

model3 = YOLO("yolo26n-pose.pt")

# 1. Use the full absolute path to the image
image_path = "images/yoga.jpeg" 

# 2. Add a quick safety check
if not os.path.exists(image_path):
    print(f"Error: File not found at {image_path}")
else:
    image1 = cv2.imread(image_path)
    
    # 3. Run prediction by explicitly passing the source image source
    results1 = model3(source=image1)

    annotated_image1 = results1[0].plot()
    plt.figure(figsize=(10,8))
    plt.imshow(cv2.cvtColor(annotated_image1, cv2.COLOR_BGR2RGB))
    plt.axis("off")
    plt.show()
```
